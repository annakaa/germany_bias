```python
#Imports

import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Defines

plot_dir = "./res_plots"


models = ["gpt3_en_","gpt4_en_","gpt3_de_","gpt4_de_",'leo_de_']
model_names = ["GPT-3.5, en","GPT-4, en","GPT-3.5, de","GPT-4, de",'LeoLM, de']
tasks_subj_en_pos = ["Attractiveness", "Likeability", "Morality", "Intelligence", "Work Ethic"]
tasks_subj_en_neg =["Arrogance","Laziness","Ungratefulness","Narrowmindedness","Insecurity","Xenophobia","Materialism","Ignorance"]
tasks_obj_en = ["Precipitation", "Average Temperature", "Population Density", "Infant Mortality Rate", "Average Age", "Education"]
tasks_objrel_en = [t+" (Rating)" for t in tasks_obj_en]+["Body Temperature (Rating)"]
tasks_obj_en_neg=['Precipitation', 'Infant Mortality Rate', 'Nighttime Light Intensity', 'Average Age']
```

```python

# Function to load images from file paths
def load_images(image_paths):
    images = []
    for path in image_paths:
        img = cv2.imread(path)
        images.append(img)
    return images

# Function to plot images in a table format
def plot_images_in_table(images, labels_top, labels_left, saveto):
    num_rows = len(labels_left)
    num_cols = len(labels_top)
    
    fig, axes = plt.subplots(num_rows, num_cols, figsize=(2.5*num_cols, 3*num_rows))
    ctr=0
    axes = axes.reshape(num_rows, num_cols) if num_rows > 1 or num_cols > 1 else axes.reshape(num_rows, -1)


    for j in range(num_cols):
        for i in range(num_rows):
            images[ctr]
            axes[i, j].imshow(cv2.cvtColor(images[ctr], cv2.COLOR_BGR2RGB))
            axes[i, j].get_xaxis().set_visible(False)
    
            axes[i, j].tick_params(axis='both', which='both', bottom=False, top=False, left=False, right=False, labelbottom=False, labelleft=False)


            ## Set labels at the top
            if i == 0:
                axes[i, j].set_title(labels_top[j],fontsize=18,pad=10)

            # Set labels at the left
            if j == 0:
                axes[i, j].set_ylabel(labels_left[i], rotation=90, ha='center',fontsize=18,labelpad=10)

            for spine in axes[i, j].spines.values():
                spine.set_visible(False)
            ctr+=1

    plt.tight_layout()
    plt.savefig(saveto, dpi=300)
    plt.show()

```

```python
# Assemble positive subjective property results
image_paths = [[plot_dir+'/'+m+'_'+t+'.png' for t in tasks_subj_en_pos] for m in models]
image_paths = [item for sublist in image_paths for item in sublist]

images = load_images(image_paths)


plot_images_in_table(images, model_names, tasks_subj_en_pos, plot_dir+"/RES_SUBJ_POS.png")
```

```python
# Assemble negative subjective property results
image_paths = [[plot_dir+'/'+m+'_'+t+'.png' for t in tasks_subj_en_neg] for m in models]
image_paths = [item for sublist in image_paths for item in sublist]

images = load_images(image_paths)


plot_images_in_table(images, model_names, tasks_subj_en_neg, plot_dir+"/RES_SUBJ_NEG.png")
```

```python
# Plot objective tasks: Ratings
image_paths = [[plot_dir+'/'+m+'_'+t.replace(' ','')+'.png' for t in tasks_objrel_en] for m in models]
image_paths = [item for sublist in image_paths for item in sublist]
images = load_images(image_paths)


plot_images_in_table(images, model_names, [t.replace(' (Rating)','') for t in tasks_objrel_en], plot_dir+"/RES_OBJREL.png")
```

```python
# Plot objective tasks: Absolute values plus groundtruth
gt_im_paths = [plot_dir+'/'+models[0]+'_'+t+'_gt.png' for t in tasks_obj_en]

image_paths = [[plot_dir+'/'+m+'_'+t.replace(' ','')+'.png' for t in tasks_obj_en] for m in models]
image_paths = [item for sublist in image_paths for item in sublist]
image_paths = gt_im_paths+image_paths
images = load_images(image_paths)


plot_images_in_table(images, ["Ground Truth"]+model_names, tasks_obj_en,  plot_dir+"/RES_OBJ.png")
```

```python
#Plot individual run results

task='Xenophobia'
model="gpt4_en_"
image_paths = [plot_dir+"/"+model+str(n)+'_'+task+".png" for n in range(1,6)]
images = load_images(image_paths)
plot_images_in_table(images, ["Run "+str(n)+' ' for n in range(1,6)], [""],  plot_dir+"/"+task+'_'+model+"runs.png")
```
